```python
df = df[["CLTV", "Contract", "Dependents", "Multiple Lines", "Offer", "Satisfaction Score", "Tenure in Months", "Total Charges", "Total Revenue", "Churn"]]
```

```python
df["Offer"] = df["Offer"].map({'Offer E':1,'Offer B':1,'Offer A':1,'Offer D':1, 'Offer C':1})
df["Contract"] = df["Contract"].map({'Two Year':24,'Month-to-Month':1,'One Year':12})
df["Offer"]=df["Offer"].replace({np.nan:0})
```
